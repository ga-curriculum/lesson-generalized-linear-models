## Generalized Linear Models (GLMs)
As we've previously discussed, **generalized linear models** describes a class of models that take the linear model (think linear regression) and generalizes the linear model beyond the assumptions of the linear regression model we discussed the other day.

Let's make this a bit more specific.

Our goal, thus far, has been to use a series of predictors $X_1,\ldots,X_p$ and use a linear combination of these predictors to learn about some $Y$ by assuming that $Y$ is linearly related to each of the $X_i$.

* When $X_i$ and $Y$ are both continuous, this is relatively easy. We can look at a plot of $X_i$ and $Y$, see if they appear approximately linearly related, and then use `statsmodels` or `scikit-learn` to tell us about how $X_i$ and $Y$ are related.

* When $Y$ is continuous but $X_i$ is discrete, it takes slightly more work, but we look at the categorical variable $X_i$ in terms of dummy variables and put that in our model. By using dummy variables, we're able to convert categories to values of `1` or `0`, which is easy for our model to understand without violating any linearity assumptions.

* When $Y$ is not continuous, however, or takes on a continuous yet finite range, we can run into some trouble. By assuming an approximately linear relationship, we're almost surely going to predict some values of $Y$ that aren't desirable. For example, if we want to use `age` to predict `will vote/won't vote`, without the coefficient for `age` being exactly 0, we will predict the probability of `will vote` to be either below 0% or above 100% for some values of `age`.

We want to still use $X_1,\ldots,X_p$ to predict $Y$. But, in cases where $Y$ is not continuous or takes on an undesirably finite range... how?

**Generalized linear models allow us to get beyond this issue.**

Let's say we write our model $Y = \mu + \varepsilon$, where $\mu$ is the "signal" and $\varepsilon$ is the "noise" accounting for the errors we do not model. We have to make some changes to $\mu$ and $\varepsilon$.

Step 1: We're going to "store" our linear combination of our predictors as $\eta = \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p$, where $\eta$ is the Greek letter "eta."

Step 2: We're going to pick some function $g(\cdot)$ that transforms the expected value of $Y$ (usually just called $\mu$) into $\eta$. Formally, we say that $g(E[Y]) = g(\mu) = \eta$. (We select some invertible function $g$ so that we can calculate $g^{-1}(\eta) = \mu = E[Y]$.
- How do you think we might pick this function?

*This takes care of $\mu$ above. Now we need to address $\varepsilon$.*

Step 3: We're going to decide what form our errors need to take. We select some probability distribution from the [exponential family of distributions](https://en.wikipedia.org/wiki/Exponential_family), which will include most of the distributions we discussed in our distributions lecture.

This sets up our GLM!

Recall that there are three components to every GLM:
* Linear Component (We still assume that a one-unit change in $X$ changes the outcome by a specific amount.)
* Link Function (How do we transform our linear component to match the desired outcome?)
* Random Component (How are the errors distributed?)

We literally did just that. Step 1 was our "linear component," where we set up the linear combination of our predictors as $\eta$. Step 2 was our "link function," where we selected some function to "bend" our linear combination of predictors to some suitable values for our outcome $Y$. Step 3 was our "random component," where we specified the distribution of errors of our model.

### Linear Regression
In linear regression, we use $Y = \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p + \varepsilon$.

In this case, our **linear component** is the $\beta_0 + \beta_1X_1 + \cdots + \beta_pX_p$. Here, $\eta = \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p$.

In this case, our **link function** is the identity function. $g(\mu) = \mu = \eta$.

In this case, our **random component** is that our errors independently follow a Normal distribution with mean 0 and fixed standard deviation $\sigma$. $\varepsilon_i \sim^\text{ind} N(0,\sigma)$.

## Logistic Regression
Let's set this up together.

* Step 1: Linear Component
<details>
$\eta = \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p$
</details>

* Step 2: Link Function
<details>
$$
\begin{eqnarray*}
E[Y] &=& \mu \\
\Rightarrow logit(\mu) &=& \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p \\
\Rightarrow \mu &=& \frac{\exp\{\beta_0 + \beta_1X_1 + \cdots + \beta_pX_p\}}{1 + \exp\{\beta_0 + \beta_1X_1 + \cdots + \beta_pX_p\}}
\end{eqnarray*}
$$
</details>

* Step 3: Random Component
<details>
$\varepsilon_i \sim Bernoulli(p_i)$
</details>

Recall that in logistic regression, we have a binary response variable coded as 0 or 1.  Formally, a logistic regression is parametrized as follows:

$Y_i \sim^\text{ind} \text{Bernoulli}(p_i)$

$g(p_i) = \mathbf{x}_i^\text{T}\beta = \beta_0 + \beta_1 x_i + \cdots + \beta_p x_p$

Where $p_i = p(\mathbf{x}_i)$ depends on the covariates $\mathbf{x}_i$, and $g$ is the _**link function**_.  The two most common logistic links are:

**Logit link:** $g(p) = \log\left(\frac{p}{1-p}\right) = \text{logit}(p)$

**Probit link:** $g(p) = \Phi^{-1}(p)$ (where $\Phi$ is the standard normal cdf.)

### The Data
The data used from this example come from one of UCLA's amazing IDRE modules (for R).  The module can be found [here](https://stats.idre.ucla.edu/r/dae/logit-regression/).

**Data Description:** _A researcher is interested in how variables, such as GRE (Graduate Record Exam scores), GPA (grade point average) and prestige of the undergraduate institution, effect admission into graduate school. The response variable, admit/don’t admit, is a binary variable._

```python
grad = pd.read_csv("https://stats.idre.ucla.edu/stat/data/binary.csv")

# Let's do some EDA!
print(grad.shape)
grad.head()
```

```python
grad.describe()
```

```python
grad.admit.value_counts()
```

```python
grad.boxplot('gre', by = 'admit')
grad.boxplot(['gpa', 'rank'], by = 'admit')
```

```python
# Now, let's build our GLM
indep_vars = ['gre', 'gpa', 'rank']
x_mat = sm.add_constant(grad[indep_vars])
y_vec = grad.admit
glm_logit = sm.GLM(y_vec, 
                   x_mat,
                   sm.families.Binomial(sm.families.links.logit)).fit()
```

```python
glm_logit.summary()
```

```python
import math
math.exp(-0.5600)
```

```python
math.exp(100*0.0023)
```

```python
# Next, let's see if we can safely reduce our model
reduced_vars = ['gre', 'gpa']
x_reduced = sm.add_constant(grad[reduced_vars])
glm_reduced = sm.GLM(y_vec,
                    x_reduced,
                    sm.families.Binomial(sm.families.links.logit)).fit()
glm_reduced.summary()
```

### Analysis of Deviance

We've spoken briefly about deviance before as a generalization of the sums of squares of error for generalized linear models.

Suppose we have two models:

$$
\begin{eqnarray*}
Y_{full} &=& \beta_0 + \beta_1X_1 + \cdots + \beta_kX_k + \cdots + \beta_pX_p + \varepsilon \\
Y_{reduced} &=& \beta_0 + \beta_1X_1 + \cdots + \beta_kX_k + \varepsilon
\end{eqnarray*}
$$

We say that $Y_{reduced}$ is nested in $Y_{full}$, because the reduced model could "fit inside" the full. (You can learn more about nested linear regression models [here](http://people.reed.edu/~jones/Courses/P24.pdf), although the ideas approximately hold for generalized linear models as well.)

When we have one model nested inside the other, there is a statistical test to see if adding new variables are worth it. (Think about it like looking at the difference in mean squared error or $R^2$ by adding a variable, but getting a $p$-value quantifying whether or not it's worth it!)

We calculate the **deviance** of the reduced model and subtract the **deviance** of the full model from it. This difference in deviance follows a [chi-squared distribution](https://en.wikipedia.org/wiki/Chi-squared_distribution) with $p-k$ degrees of freedom. (Note that $p-k$ indicates how many variables we took out of our full model to get to the reduced model!)

**This comparison only works with nested models! Do not use if your models are not nested!**

```python
# Test model differences
from scipy.stats import chi2

D = glm_reduced.deviance - glm_logit.deviance
print('Difference in Deviance: ', D)
pval = 1 - chi2.cdf(D, df = 1)
print('p-value of test of difference: ', pval) # What can we conclude here?
```

### Poisson Regression Example
Let's set this up together.

* Step 1: Linear Component
<details>
$\eta = \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p$
</details>

* Step 2: Link Function
<details>
$$
\begin{eqnarray*}
E[Y] &=& \mu \\
\Rightarrow log(\mu) &=& \beta_0 + \beta_1X_1 + \cdots + \beta_pX_p \\
\Rightarrow \mu &=& \exp\{\beta_0 + \beta_1X_1 + \cdots + \beta_pX_p\}
\end{eqnarray*}
$$
</details>

* Step 3: Random Component
<details>
$\varepsilon_i \sim Poisson(\lambda_i)$
</details>

Formal definition of a Poisson regression:

$Y_i \sim^\text{ind} \text{Poisson}(\lambda_i)$

$\log (\lambda_i) = \mathbf{x}_i^\text{T}\beta = \beta_0 + \beta_1 x_1 + \cdots + \beta_p x_p$

where, again, $\lambda_i = \lambda(\mathbf{x}_i)$ depends on $\mathbf{x}_i$.

#### Data
We'll again rely on UCLA's IDRE module.  This one can be found [here](https://stats.idre.ucla.edu/r/dae/poisson-regression/).

#### Data Description
_The number of awards earned by students at one high school. Predictors of the number of awards earned include the type of program in which the student was enrolled (e.g., vocational, general or academic) and the score on their final exam in math._

```python
award = pd.read_csv("https://stats.idre.ucla.edu/stat/data/poisson_sim.csv")

# Let's do some EDA:
award.head()
```

```python
award.plot('math', 'num_awards', kind = 'scatter')
award.boxplot('num_awards', by = 'prog')
```

```python
# Notice that prog is actually a categorical variable - I am aware of this.
# I'm going to suspsend that knowledge for the sake of example.
poi_vars = ['prog', 'math']
x_poi = sm.add_constant(award[poi_vars])
y_poi = award.num_awards

glm_poi = sm.GLM(y_poi,
                 x_poi,
                 family = sm.families.Poisson()).fit()
glm_poi.summary()
```

```python
# Interesting - it looks like prog is pretty insignificant.
# Let's try removing it and seeing what happens.
x_poi_red = sm.add_constant(award.math)

glm_poi_red = sm.GLM(y_poi,
                     x_poi_red,
                     family = sm.families.Poisson()).fit()
glm_poi_red.summary()
```

```python
# Test model difference:
D_poi = glm_poi_red.deviance - glm_poi.deviance
print('Difference in Deviance: ', D_poi)
pval_poi = 1 - chi2.cdf(D_poi, 1)
print('p-value of test of difference: ', pval_poi) # What does this mean?
```
